```python
import pickle, json, re
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit import RDLogger
RDLogger.DisableLog('rdApp.*')
import pyarrow as pa

RESULTS    = "/data/vu/retromech/baselines/nerf/results"
MAELLE_DIR = "/home/vu/repos/retromech/octavian_investigation/uspto_480k_unified/maelle/inference_59450"

# per-split file registry (add more splits here as their aligned-input files are built)
SPLITS = {
    "ood_mass": {
        "nerf":    f"{RESULTS}/uspto480k_unified_ood_mass_0.0_2019.pickle",
        "aligned": "/home/vu/repos/NERF/results/ood_mass_aligned_input.pickle",
        "maelle":  f"{MAELLE_DIR}/test_ood_mass_8_steps_t_pois_06_t_cat_1_do_sampling/data-00000-of-00001.arrow",
    },
    "ood_ester": {
        "nerf":    f"{RESULTS}/uspto480k_unified_ood_ester_0.0_2019.pickle",
        "aligned": "/home/vu/repos/NERF/results/ood_ester_aligned_input.pickle",
        "maelle":  f"{MAELLE_DIR}/test_ood_ester_8_steps_t_pois_06_t_cat_1_do_sampling/data-00000-of-00001.arrow",
    },
}

_cache = {}
def _load(split):
    if split not in _cache:
        c = SPLITS[split]
        nerf = pickle.load(open(c["nerf"], "rb"))["pred"]          # pred[k], pred/pickle order
        aligned = pickle.load(open(c["aligned"], "rb"))            # aligned[k] -> {hf_idx, reactant, product_main, ...}
        with pa.memory_map(c["maelle"], "r") as s:                 # Maelle in HF order -> index by hf_idx
            counts = pa.ipc.open_stream(s).read_all().column("counts").to_pylist()
        _cache[split] = (nerf, aligned, counts)
    return _cache[split]

_maps = re.compile(r":\d+")
def _mol(smi):                       # parse + strip atom maps for clean drawing
    if not smi: return None
    return Chem.MolFromSmiles(_maps.sub("", smi))
def _canon(smi):
    m = _mol(smi); return Chem.MolToSmiles(m) if m else None
def _nheavy(smi):
    m = Chem.MolFromSmiles(smi); return m.GetNumHeavyAtoms() if m else -1

def show_case(k, split="ood_mass"):
    """Visualize NERF prediction index k vs Maelle for the same reaction in `split`."""
    nerf, aligned, counts = _load(split)
    inp = aligned[k]; hf = inp["hf_idx"]
    truth = _canon(inp["product_main"])

    nerf_set  = nerf[k]                                             # list of fragments or None
    nerf_main = max(nerf_set, key=_nheavy) if nerf_set else None    # largest fragment = main product
    ranked    = sorted(json.loads(counts[hf]).items(), key=lambda kv: -kv[1])
    maelle_top1, maelle_cnt = ranked[0]

    nerf_ok   = _canon(nerf_main)   == truth
    maelle_ok = _canon(maelle_top1) == truth

    print(f"[{split}] pred index k={k}  ->  HF row {hf}")
    print("reactant    :", _canon(inp["reactant"]))
    print("true product:", truth)
    print(f"NERF   top-1: {_canon(nerf_main)}   [{'HIT' if nerf_ok else 'MISS'}]")
    print(f"       full set: {None if nerf_set is None else '.'.join(nerf_set)}")
    print(f"MAELLE top-1: {_canon(maelle_top1)}   [{'HIT' if maelle_ok else 'MISS'}]  ({maelle_cnt} samples)")
    print("MAELLE top-5:", [(_canon(s), c) for s, c in ranked[:5]])

    mols    = [_mol(inp["reactant"]), _mol(truth), _mol(nerf_main), _mol(maelle_top1)]
    legends = ["reactant (input)", "TRUE product",
               f"NERF top-1 [{'HIT' if nerf_ok else 'MISS'}]",
               f"MAELLE top-1 [{'HIT' if maelle_ok else 'MISS'}]"]
    mols = [m if m is not None else Chem.MolFromSmiles("") for m in mols]   # placeholder for invalid
    return Draw.MolsToGridImage(mols, legends=legends, molsPerRow=2, subImgSize=(450, 350))

show_case(898, "ood_mass")
```

```python
# same visualization for the OOD ester split (ester dropped 0 rows, so hf_idx == k)
random_index = random.randint(0, 1800 - 1)
print(f"Random index: {random_index}")
show_case(random_index, "ood_ester")
```
