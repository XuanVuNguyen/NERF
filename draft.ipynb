```python
import pickle

with open("/data/vu/retromech/baselines/nerf/results/0.7_2019.pickle", "rb") as f:
    r = pickle.load(f)

print(r["temperature"])  # 0.0  (the sampling temperature)
print(r["acc"])          # top-1 exact-match accuracy on the test set
print(r["valid"])        # fraction of predictions that are chemically valid
print(r["ckpt"])         # list of checkpoint(s) that were averaged

# r["pred"] is one entry per test reaction (in test-set order):
for pred in r["pred"][:5]:
    if pred is None:
        print("invalid prediction")
    else:
        for each_pred in pred:
            print(each_pred)
    
    print('=====')
```
